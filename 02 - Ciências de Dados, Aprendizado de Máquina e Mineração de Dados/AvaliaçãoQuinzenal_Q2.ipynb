### Questão 1
Dadas as seguintes matrizes, calcule as medidas de avaliação baseadas na matriz de confusão (TFP, TFN, TVN, TVP).

Matriz de confusão do classificador '1' <br>
[[41 59]  <br>
 [ 3 97]] <br>
Matriz de confusão do classificador '2' <br>
[[96  4]  <br>
 [ 5 95]]  <br>
Matriz de confusão do classificador '3'   <br>
[[98  2]  <br>
 [ 3 97]]  <br>
e escolha a **alternativa correta**. <br>

a)	A precisão obtida pelo Classificador '1' foi maior que para o Classificador '2'.<br>
b)	As acurácias obtidas pelos três classificadores foram acima de 0.90 . <br>
c)	Pode-se dizer que o Classificador '3' foi melhor que os demais, pois apresentou maior valor em todas as métricas calculadas.  <br>
d)	O valor do F1-score foi maior para Classificador '1' do que para a Classificador '3'. <br>
e)	Nenhuma das anteriores. <br>

```python
# Código executado para mostrar as métricas (TFP, TFN, TVN, TVP, Precisão, Acurácia, F1)
import numpy as np
import pandas as pd
from math import isclose

# Matrizes fornecidas (formato: [[TN, FP], [FN, TP]])
cms = {
    '1': np.array([[41, 59],
                   [ 3, 97]]),
    '2': np.array([[96, 4],
                   [ 5, 95]]),
    '3': np.array([[98, 2],
                   [ 3, 97]])
}

rows = []
for key, cm in cms.items():
    TN, FP = int(cm[0,0]), int(cm[0,1])
    FN, TP = int(cm[1,0]), int(cm[1,1])
    total = TN + FP + FN + TP
    
    # métricas (TFP = taxa de falso positivo (FP/(FP+TN)); TFN = taxa de falso negativo (FN/(FN+TP))
    TFP = FP / (FP + TN)
    TFN = FN / (FN + TP)
    TVN = TN / (TN + FP)   # taxa verdadeira negativo / especificidade
    TVP = TP / (TP + FN)   # taxa verdadeira positivo / recall
    Precision = TP / (TP + FP) if (TP + FP) > 0 else 0
    Accuracy = (TP + TN) / total
    Recall = TVP
    F1 = 2 * Precision * Recall / (Precision + Recall) if (Precision + Recall) > 0 else 0
    
    rows.append({
        'Classifier': key,
        'TN': TN, 'FP': FP, 'FN': FN, 'TP': TP, 'Total': total,
        'TFP': TFP, 'TFN': TFN, 'TVN': TVN, 'TVP': TVP,
        'Precision': Precision, 'Accuracy': Accuracy, 'F1': F1,
        'TFP_frac': f"{FP}/{FP+TN}",
        'TFN_frac': f"{FN}/{FN+TP}",
        'TVN_frac': f"{TN}/{TN+FP}",
        'TVP_frac': f"{TP}/{TP+FN}",
        'Precision_frac': f"{TP}/{TP+FP}",
        'Accuracy_frac': f"{TP+TN}/{total}"
    })

df = pd.DataFrame(rows).set_index('Classifier')

# Mostrar tabela com três casas decimais nas métricas principais
display_df = df[['TN','FP','FN','TP','Total',
                 'TFP','TFN','TVN','TVP',
                 'Precision','Accuracy','F1']].copy()
display_df = display_df.round({
    'TFP':3,'TFN':3,'TVN':3,'TVP':3,'Precision':3,'Accuracy':3,'F1':3
})

# Apresenta a tabela
try:
    # exibe como tabela interativa caso a função esteja disponível
    import caas_jupyter_tools as cjt
    cjt.display_dataframe_to_user("Métricas por classificador", display_df)
except Exception:
    print(display_df)

# Verificação das alternativas pedidas
prec1 = df.loc['1','Precision']
prec2 = df.loc['2','Precision']
prec3 = df.loc['3','Precision']

acc1 = df.loc['1','Accuracy']
acc2 = df.loc['2','Accuracy']
acc3 = df.loc['3','Accuracy']

f1_1 = df.loc['1','F1']
f1_2 = df.loc['2','F1']
f1_3 = df.loc['3','F1']

# (a) Precisão 1 > Precisão 2 ?
opt_a = prec1 > prec2

# (b) acurácias dos três > 0.90 ?
opt_b = (acc1 > 0.90) and (acc2 > 0.90) and (acc3 > 0.90)

# (c) Classificador 3 tem MAIOR valor em TODAS as métricas calculadas?
# interpretamos "maior valor" como estritamente maior numericamente para cada métrica
metrics = ['TFP','TFN','TVN','TVP','Precision','Accuracy','F1']
opt_c = all(df.loc['3',m] > df.loc['1',m] and df.loc['3',m] > df.loc['2',m] for m in metrics)

# (d) F1 do 1 > F1 do 3 ?
opt_d = f1_1 > f1_3

# Resultado final
print("\nComparações (True = alternativa VERDADEIRA):")
print(f"(a) Precisão(1) > Precisão(2)? --> {opt_a}   (prec1={prec1:.6f}, prec2={prec2:.6f})")
print(f"(b) Todas acurácias > 0.90? --> {opt_b}   (acc1={acc1:.6f}, acc2={acc2:.6f}, acc3={acc3:.6f})")
print(f"(c) Classif.3 tem MAIOR valor em TODAS métricas? --> {opt_c}")
print(f"    (valores resumidos: Classif3 vs Classif1 vs Classif2)")
for m in metrics:
    print(f"    {m}: {df.loc['3',m]:.6f} vs {df.loc['1',m]:.6f} vs {df.loc['2',m]:.6f}")
print(f"(d) F1(1) > F1(3)? --> {opt_d}   (F1_1={f1_1:.6f}, F1_3={f1_3:.6f})")

# Conclusão lógica
if not (opt_a or opt_b or opt_c or opt_d):
    conclusion = "Alternativa correta: (e) Nenhuma das anteriores."
else:
    true_opts = []
    if opt_a: true_opts.append('(a)')
    if opt_b: true_opts.append('(b)')
    if opt_c: true_opts.append('(c)')
    if opt_d: true_opts.append('(d)')
    conclusion = "Alternativa(s) verdadeira(s): " + ", ".join(true_opts)

print("\n" + conclusion)
```
